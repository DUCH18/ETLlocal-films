```python
import json
import pandas as pd
from pathlib import Path


path = '../data/raw/latest_films.json' 
with open(path) as f:
    data = json.load(f)
        

df = pd.json_normalize(data["results"])
        
df.head()
```

```python
cols_to_drop = [
    "adult",
    "backdrop_path",
    "original_title",
    "poster_path"
]

d = df.drop(cols_to_drop, axis=1)
d["release_date"] = pd.to_datetime(d["release_date"])
d = d.sort_values(by="release_date")
d.head()
```

```python
import pandas as pd
import sqlite3

# load csv
df = pd.read_csv("data.csv")

# create in-memory database
conn = sqlite3.connect(":memory:")

# send dataframe to SQL
df.to_sql("table1", conn, index=False)

# run SQL query
query = """
SELECT name, age
FROM table1
WHERE age > 25
"""

result = pd.read_sql(query, conn)

print(result)
```
